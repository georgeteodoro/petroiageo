### Número total de pontos preditos a cada iteração

```python
num_predicted = list()
for i in range(1, 281):
    num_predicted.append(cube.num_predicted_at_it(i))
```

```python
from matplotlib.animation import FuncAnimation
from IPython import display
```

```python
y_max = 433
x_max = 645

Figure = plt.figure(figsize=(10, 6))
image = plt.imshow(np.random.random((x_max, y_max)))
plt.colorbar()

plt.xlim(0, x_max)
plt.ylim(0, y_max)

def init_function():
    image.set_array(np.random.random((x_max, y_max)))
    return [image]

def AnimationFunction(frame):
 
    a = image.get_array()
    a = cube.pred_config_at_it(frame)
    image.set_array(a)
    return [image]

anim_created = FuncAnimation(Figure, AnimationFunction, frames = list(range(1, 281)), interval = 25, init_func=init_function, blit=True)
```

```python
video = anim_created.to_html5_video()
html = display.HTML(video)
display.display(html)
 
# good practice to close the plt object.
plt.close()
```
